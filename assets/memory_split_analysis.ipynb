```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
// > memory.txt is expected to hold the ouput of "./gradlew runSplitMemoryMeasurement"
val memory = DataFrame.read("split_memory.csv")
```

```kotlin
private enum class CollectionType(val text: String, val prettyName: String, val sizeMultiplier: Int = 1) {
    LIST("list", "List"),
    SET("set", "Set"),
    MAP("map", "Map", 2)
}

private val elementsSize = 1000000 * 4 // benchmark ran against 1 million ints
private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .take(100)
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("numCollections").cast<Int>().with { "${it} x ${1000000 / it}" }
    .convert("totalBytes").cast<Int>().with { (it - collectionType.sizeMultiplier*elementsSize) / (1024.0 * 1024.0) }
    .plot {
        bars {
            x("numCollections") {
                axis.name = "Collection Split (# of ${collectionType.prettyName}s x # of Elements)"
            }
            y("totalBytes") {
                axis.name = "Overhead (MB)"
            }
            fillColor("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "Eclipse" to Color.GREY,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "1 Million Elements ${collectionType.prettyName} Split Overhead"
            size = 894 to 800
        }
    }
```

```kotlin
private enum class CollectionType(val text: String, val prettyName: String)  {
    LIST("emptylist", "List"),
    SET("emptyset", "Set"),
    MAP("emptymap", "Map")
}

private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .take(100)
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("numCollections").cast<Int>().with { "${it} x ${1000000 / it}" }
    .convert("totalBytes").cast<Int>().with { it / (1024.0 * 1024.0) }
    .plot {
        bars {
            x("numCollections") {
                axis.name = ""
                axis.breaks(emptyList())
            }
            y("totalBytes") {
                axis.name = "Overhead (MB)"
            }
            fillColor("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "Eclipse" to Color.WHITE,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "1 Million Empty ${collectionType.prettyName}s Overhead"
            size = 894 to 800
        }
    }
```
