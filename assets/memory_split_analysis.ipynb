```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
// > memory.txt is expected to hold the ouput of "./gradlew runSplitMemoryMeasurement"
val memory = DataFrame.read("split_memory.txt")
```

```kotlin
private enum class CollectionType(val text: String, val sizeMultiplier: Int = 1) {
    LIST("list"),
    SET("set"),
    MAP("map", 2)
}

private val elementsSize = 1000000 * 4 // benchmark ran against 1 million ints
private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .take(100)
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("numCollections").cast<Int>().with { "${it} x ${1000000 / it}" }
    .convert("totalBytes").cast<Int>().with { (it - collectionType.sizeMultiplier*elementsSize) / (1024.0 * 1024.0) }
    .plot {
        bars {
            x("numCollections") {
                axis.name = "Collection Split (# of Collections x # of Elements)"
            }
            y("totalBytes") {
                axis.name = "Overhead (MB)"
            }
            fillColor("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "1 Million Elements Collection Split Overhead"
            size = 894 to 800
        }
    }
```

```kotlin
private enum class CollectionType(val text: String) {
    LIST("emptylist"),
    SET("emptyset"),
    MAP("emptymap")
}

private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .take(100)
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("numCollections").cast<Int>().with { "${it} x ${1000000 / it}" }
    .convert("totalBytes").cast<Int>().with { it / (1024.0 * 1024.0) }
    .plot {
        bars {
            x("numCollections") {
                axis.name = ""
                axis.breaks(emptyList())
            }
            y("totalBytes") {
                axis.name = "Overhead (MB)"
            }
            fillColor("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "1 Million Empty Collections Overhead"
            size = 894 to 800
        }
    }
```
