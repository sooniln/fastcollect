```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
val memory = DataFrame.read("memory.txt")

// overall graph
val collectionType = "map"
memory
    .filter { "collection"<String>().startsWith(collectionType) }
    .plot {
        line {
            x("size")
            y("bytes")
            color("collection") {
                scale = categorical(
                    "${collectionType}FastCollect" to Color.RED,
                    "${collectionType}Fastutil" to Color.BLUE,
                    "${collectionType}Jvm" to Color.GREEN,
                    "${collectionType}Kds" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "Collection Memory Usage"
            size = 900 to 550
        }
    }
```

```kotlin
val memory = DataFrame.read("memorySplit.txt")

// overall graph
val collectionType = "map"
val rawCollectionSize = 2 * 4 * 1000000

memory
    .filter { "collection"<String>().startsWith(collectionType) }
    .convert("numCollections") { "${1000000 / it as Int}x${it}" }
    .update("totalBytes").cast<Int>().with { it - rawCollectionSize }
    .plot {
        line {
            x("numCollections")
            y("totalBytes")
            color("collection") {
                scale = categorical(
                    "${collectionType}FastCollect" to Color.RED,
                    "${collectionType}Fastutil" to Color.BLUE,
                    "${collectionType}Jvm" to Color.GREEN,
                    "${collectionType}Kds" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "Collection Overhead"
            size = 900 to 550
        }
    }

memory
    .filter { "collection"<String>().startsWith("empty${collectionType}") }
    .convert("numCollections").to<String>()
    .plot {
        bars {
            x("numCollections")
            y("totalBytes")
            fillColor("collection") {
                scale = categorical(
                    "empty${collectionType}FastCollect" to Color.RED,
                    "empty${collectionType}Fastutil" to Color.BLUE,
                    "empty${collectionType}Jvm" to Color.GREEN,
                    "empty${collectionType}Kds" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "Empty Collection Overhead"
            size = 900 to 550
        }
    }
```
