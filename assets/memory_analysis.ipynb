```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
// > memory.txt is expected to hold the ouput of "./gradlew runMemoryMeasurement"
val memory = DataFrame.read("memory.txt")
```

```kotlin
private enum class CollectionType(val text: String) {
    LIST("list"),
    SET("set"),
    MAP("map")
}

private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("bytes").cast<Int>().with { it / (1024.0 * 1024.0) }
    .plot {
        line {
            x("size") {
                axis.name = "Collection Size"
            }
            y("bytes") {
                axis.name = "Memory (MB)"
            }
            color("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "Collection Memory Usage"
            size = 894 to 800
        }
    }
```
