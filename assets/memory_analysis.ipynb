```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
// > memory.csv is expected to hold the ouput of "./gradlew runMemoryMeasurement"
val memory = DataFrame.read("memory.csv")
```

```kotlin
import org.jetbrains.letsPlot.core.spec.plotson.scale

private enum class CollectionType(val text: String, val prettyName: String) {
    LIST("list", "List"),
    SET("set", "Set"),
    MAP("map", "Map")
}

private val collectionType = CollectionType.MAP

memory
    .filter { "collection"<String>().startsWith(collectionType.text) }
    .update("collection").cast<String>().with { it.removePrefix(collectionType.text) }
    .convert("bytes").cast<Int>().with { it / (1024.0 * 1024.0) }
    .plot {
        line {
            x("size") {
                axis.name = "Collection Size"
                scale = continuous(min = 0)
            }
            y("bytes") {
                axis.name = "Memory (MB)"
                scale = continuous(min = 0)
            }
            color("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "Eclipse" to Color.WHITE,
                    "AndroidX" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
            type = LineType.DASHED
        }
        points {
            x("size")
            y("bytes")
            size = 2.0
            color("collection") {
                legend.name = ""
                scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "Eclipse" to Color.WHITE,
                    "AndroidX" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )
            }
        }
        layout {
            title = "${collectionType.prettyName} Memory Usage"
            size = 894 to 800
        }
    }
```
