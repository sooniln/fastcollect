```kotlin
// > memory.txt is expected to hold the ouput of "./gradlew runMemoryMeasurement"
val cpu = DataFrame.read("set_putmiss.csv")
```

```kotlin
cpu
    .filter { "library"<String>() != "eclipse" && "library"<String>() != "hashsmith" }
    .plot {
        line {
            x("size") {
                axis.name = "# of Elements (log scale)"
                scale = continuous(transform = Transformation.LOG10, min = 3000)
            }
            y("score") {
                axis.name = "Time (log scale)"
                axis.breaksLabeled()
                scale = continuous(transform = Transformation.LOG10)
            }
            color("library") {
                legend.name = ""
                /*scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "HashSmith" to Color.YELLOW,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )*/
            }
        }
        layout {
            title = "Int Set PutMiss Benchmark"
            size = 800 to 400
        }

        vLine {
            xIntercept.constant(8000)
            type = LineType.DASHED
            width = 1.0
        }
        text {
            x.constant(8000)
            y.constant(20)
            label = "L1 Cache"
        }

        vLine {
            xIntercept.constant(250000)
            type = LineType.DASHED
            width = 1.0
        }
        text {
            x.constant(250000)
            y.constant(20)
            label = "L2 Cache"
        }
    }
```
