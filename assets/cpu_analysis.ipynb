```kotlin
%useLatestDescriptors
%use dataframe
%use kandy
```

```kotlin
// > memory.txt is expected to hold the ouput of "./gradlew runMemoryMeasurement"
val cpu = DataFrame.read("benchmark_results_2026-05-05.csv")
```

```kotlin
private enum class BenchmarkType(val collection: String, val operation: String) {
    LIST_GROW("List", "Grow"),
    LIST_ITERATE("List", "Iterate"),
    LIST_SEARCH("List", "Searc"),
    SET_GETHIT("Set", "GetHit"),
    SET_GETMISS("Set", "GetMiss"),
    SET_PUTHIT("Set", "PutHit"),
    SET_PUTMISS("Set", "PutMiss"),
    SET_GROW("Set", "Grow"),
    SET_ITERATE("Set", "Iterate"),
    MAP_GETHIT("Map", "GetHit"),
    MAP_GETMISS("Map", "GetMiss"),
    MAP_PUTHIT("Map", "PutHit"),
    MAP_PUTMISS("Map", "PutMiss"),
    MAP_GROW("Map", "Grow"),
    MAP_ITERATE("Map", "Iterate"),
    MAP_ITERATE_FAST("Map", "IterateFast"),
    MAP_ITERATE_KEYS("Map", "IterateKeys"),
    MAP_ITERATE_VALUES("Map", "IterateValues"),
}

private val benchmarkType = BenchmarkType.LIST_ITERATE

cpu
    .filter { "benchmark"<String>().equals(benchmarkType.collection) && "operation"<String>().equals(benchmarkType.operation) }
    .filter { "size"<Int>().equals(1536000) }
    .convert("size").cast<Int>().with { it.toString() }
    .plot {
        bars {
            x("size") {
                axis.name = ""
            }
            y("score") {
                axis.name = "Time"
                axis.breaks(listOf())
                scale = continuous(min = 0)
            }
            fillColor("implementation") {
                legend.name = ""
                /*scale = categorical(
                    "FastCollect" to Color.RED,
                    "Fastutil" to Color.BLUE,
                    "HashSmith" to Color.YELLOW,
                    "KDS" to Color.GREEN,
                    "Kotlin" to Color.PURPLE,
                )*/
            }
        }
        layout {
            title = "${benchmarkType.collection} ${benchmarkType.operation} Operation Time"
            size = 894 to 800
        }
    }
```
